```python
import numpy as np
```

```python
#1. Creating two linear arrays `a` and `b` with 20 random elements each ranging from 1 to 100.
a = np.random.randint(1, 101, size=20)
b = np.random.randint(1, 101, size=20)
print("Array a:", a)
print("Array b:", b)
```

```python
# 2. Changing those linear arrays to two 5x4 matrices.
a = a.reshape(5, 4)
b = b.reshape(5, 4)
print("\nMatrix a:\n", a)
print("\nMatrix b:\n", b)
```

```python
# 3. Scalar multiplication with matrix `a` and `b`.
scalar_a = 2 * a  # Scalar multiplication by 2
scalar_b = 3 * b  # Scalar multiplication by 3
print("\nScalar multiplication of matrix a by 2:\n", scalar_a)
print("\nScalar multiplication of matrix b by 3:\n", scalar_b)
```

```python
# 4. Modify matrix `b`, reshape it to 4x5, and perform matrix multiplication.
b = b.reshape(4, 5)  # Reshape `b` to 4x5 for matrix multiplication
c = np.dot(a, b)  # Matrix multiplication of `a` (5x4) and reshaped `b` (4x5)
print("\nMatrix c (result of matrix multiplication of a and reshaped b):\n", c)
```

```python
# 5. Printing only columns 2 and 3 of matrix `c` using slicing.
print("\nColumns 2 and 3 of matrix c:\n", c[:, 1:3])

# Printing the entire matrix `c` using slicing technique.
print("\nEntire matrix c:\n", c[:, :])
```

```python
# 6. Find the largest element in `c` and the index of that element.
max_element = np.max(c)  # Find the largest element in `c`
max_index = np.unravel_index(np.argmax(c), c.shape)  # Convert flat index to 2D index
print("\nLargest element in matrix c:", max_element)
print("Index of the largest element:", max_index)  # Returns (row, column) format
```

```python
# 7. Converting the matrix `c` into a one-dimensional array without modifying the original.
c_flat = c.flatten()
print("\nMatrix c as a one-dimensional array:\n", c_flat)
```
